```python
from langchain.llms.huggingface import HuggingFace
from langchain.llms.openai import OpenAIChat
```

```python
openai_key = ''
model = OpenAIChat(
    model_name='gpt-4o',
    batch_size=1,
    openai_api_key=openai_key,
    request_timeout=1000, base_delay=10, keep_trying=True,
    frequency_penalty=0,
    presence_penalty=0
)

def format_prompts(prompts):
    messages = []
    for i, j in prompts:
        messages.append([dict(
            role='user', content=i,
        ),
        dict(
            role='user', content=j,
        )])
        
    return messages
```

```python
titles = ['Exiled from The Black Community! This video of me OnStage pissed the left off. Want me to Remove it.',
 "Gutfeld on Jussie Smollett's arrest",
 "President Reagan's Remarks to the National Rifle Association on May 6, 1983",
 'Ben Carson HUMILIATES AOC With An Epic Speech, Gets A Standing Ovation',
 'Nigeria destroys more than 1 million AstraZeneca COVID vaccine doses | Omicron Variant | Health',
 'Greg Gutfeld: Something has got to be done about Adam Schiff',
 "Rittenhouse prosecutors' evidence has 'disintegrated': Gregg Jarrett",
 'WOKE Girl Claims She Feels Miserable With Her Life Because Everything Is Racist & Microaggressions',
 '“This Is When It Matters.” | Chet Huntley | MSNBC',
 "Lavrov on Comey: 'You're kidding!'"]
```

```python
prompts = [
(
    'Classify the following YouTube video title as ideologically liberal, ideologically neutral, or ideologically conservative. Titles with no ideological content are classified as neutral.',
    f'Title: {title}'
) for title in titles
]
```

```python
output = model.generate(format_prompts(prompts))
```

```python
for generation in output.generations:
    print(generation[0].text)
    print(''.ljust(80, '='))
```

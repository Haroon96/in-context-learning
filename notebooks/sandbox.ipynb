```python
import pandas as pd
import json
from pathlib import Path
```

```python
DATA_DIR = Path('/home/mharoon/in-context-learning/data/classification/news_ideology/Article-Bias-Prediction-main/data/')
```

```python
train = pd.read_csv(DATA_DIR / 'splits/media/train.tsv', delimiter='\t')['ID']
test = pd.read_csv(DATA_DIR / 'splits/media/test.tsv', delimiter='\t')['ID']
```

```python
OUTPUT_DIR = Path('/home/mharoon/in-context-learning/data/classification/news_ideology/')
```

```python
train_set = []
for _id in train:
    with open(DATA_DIR / f'jsons/{_id}.json') as f:
        js = json.load(f)
    train_set.append(js)
train_df = pd.DataFrame(train_set)
```

```python
test_set = []
for _id in test:
    with open(DATA_DIR / f'jsons/{_id}.json') as f:
        js = json.load(f)
    test_set.append(js)
test_df = pd.DataFrame(test_set)
```

```python
train_df.to_csv(OUTPUT_DIR / 'train.csv', index=False)
```

```python
test_df.to_csv(OUTPUT_DIR / 'test.csv', index=False)
```

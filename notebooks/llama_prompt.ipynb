```python
from langchain.llms.huggingface import HuggingFace
import pandas as pd
```

```python
generation_kwargs = dict(
    temperature=0.0, 
    max_tokens=300, 
    top_p=1.0
)

common_kwargs = dict(
    model_name='llama-7B',
    batch_size=1,
    verbose=True
)

generation_kwargs['max_new_tokens'] = generation_kwargs.pop('max_tokens')
model = HuggingFace.from_model_name(
**common_kwargs, task='text-generation', device=1, cache=False,
generation_kwargs=generation_kwargs | dict(do_sample=False))
```

```python
test = pd.read_csv('../data/classification/yt_ideology/test.csv')
titles = test[test['label'].isin([0, 2])].sample(10, random_state=10)['text']
titles
```

```python
prompts = [
'''
Classify the following YouTube video title as ideologically liberal, ideologically neutral, or ideologically conservative. Titles with no ideological content are classified as neutral. 
Title: %s
Explain step-by-step how the classification was made.
'''.strip() % title for title in titles
]
```

```python
output = model.generate(prompts)
```

```python
for generation in output.generations:
    print(generation[0].text)
    print(''.ljust(80, '='))
```
